```python
roman = {'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100, 'D': 500, 'M': 1000}
```

```python
s = 'MCMXCIV'
```

```python
sum = 0

for i in range(len(s) - 1):
    print(f"Character right now: {s[i]}\tNext character: {s[i+1]}")
    if roman[s[i]] < roman[s[i+1]]:
        print(f"Since {roman[s[i]]} < {roman[s[i+1]]}, subtracting {roman[s[i]]} from sum.")
        sum -= roman[s[i]]
        print(f"Sum right now: {sum}")
    else:
        print(f"Since {roman[s[i]]} >= {roman[s[i+1]]}, adding {roman[s[i]]} to sum.")
        sum += roman[s[i]]
        print(f"Sum right now: {sum}")

sum += roman[s[-1]]
print(f"Sum right now: {sum}")
```
