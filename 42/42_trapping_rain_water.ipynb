```python
import tracemalloc

def trap_original(height):
    if not height or len(height) < 3:
        return 0

    n = len(height)
    left_max = [0] * n
    right_max = [0] * n
    reversed_height = height[::-1]  # Extra memory usage

    trapped_water = [0] * n  # Extra array

    left_max[0] = height[0]
    right_max[0] = reversed_height[0]

    for i in range(1, n):
        left_max[i] = max(left_max[i - 1], height[i])
        right_max[i] = max(right_max[i - 1], reversed_height[i])

    right_max = right_max[::-1]  # Extra memory usage

    for i in range(n):
        trapped_water[i] = min(left_max[i], right_max[i]) - height[i]

    return sum(trapped_water)

def trap_optimized(height):
    if not height or len(height) < 3:
        return 0

    n = len(height)
    left_max = [0] * n
    right_max = [0] * n
    trapped_water = 0  # No extra array

    left_max[0] = height[0]
    for i in range(1, n):
        left_max[i] = max(left_max[i - 1], height[i])

    right_max[n - 1] = height[n - 1]
    for i in range(n - 2, -1, -1):
        right_max[i] = max(right_max[i + 1], height[i])

    for i in range(n):
        trapped_water += min(left_max[i], right_max[i]) - height[i]

    return trapped_water
```

```python
# Sample input
height = [0,1,0,2,1,0,1,3,2,1,2,1]
```

```python
%%timeit
# Measure memory for optimized approach
tracemalloc.start()
trap_optimized(height)
mem_optimized = tracemalloc.get_traced_memory()
tracemalloc.stop()
```

```python
%%timeit
# Measure memory for original approach
tracemalloc.start()
trap_original(height)
mem_original = tracemalloc.get_traced_memory()
tracemalloc.stop()
```

```python

# Print memory usage
print(f"Original Approach: {mem_original[1] / 1024:.4f} KB (Peak)")
print(f"Optimized Approach: {mem_optimized[1] / 1024:.4f} KB (Peak)")
```

```python
left = 0
right = len(height) - 1
left_max = height[left]
right_max = height[right]
trapped_water = 0

while left < right:
    print(f"\nCurrent left: {left}, right: {right}")
    if left_max < right_max:
        print(f"Left_max is less than right_max. Left_max: {left_max}, right_max: {right_max}")
        print(f"Height at left position {left}: {height[left]}")
        left += 1
        print(f"Index moved to position: {left}.\tHeight at new left position: {height[left]}")
        left_max = max(left_max, height[left])
        print(f"Updated left_max to {left_max}.")
        trapped_water += max(0, left_max - height[left])
        print(f"Max of 0 and ({left_max} - {height[left]}) is {max(0, left_max - height[left])}.")
    else:
        print(f"Left_max is not less than right_max. Left_max: {left_max}, right_max: {right_max}")
        print(f"Height at right position {right}: {height[right]}")
        right -= 1
        print(f"Index moved to position: {right}.\tHeight at new right position: {height[right]}")
        right_max = max(right_max, height[right])
        print(f"Updated right_max to {right_max}.")
        trapped_water += max(0, right_max - height[right])
        print(f"Max of 0 and ({right_max} - {height[right]}) is {max(0, right_max - height[right])}.")
    print(f"Total trapped water: {trapped_water}")

# print(f"\n\nTotal trapped water: {trapped_water}")
```
